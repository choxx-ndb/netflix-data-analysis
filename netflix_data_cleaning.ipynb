# Netflix Data Analysis Project
Exploratory data analysis of Netflix titles dataset using Python, Pandas, and Matplotlib.

## Project Overview
This project explores the Netflix titles dataset to understand content distribution, ratings, countries, genres ,and movie durations .
The workflow includes data cleaning, preparation ,and exploratory data analysis.

## Objectives
- Clean and prepare the dataset
- Explore the distribution of movies vs TV Shows
- Analyze trends over time
- Identify top countries, ratings, and genres
- Study movie duration patterns

```python
# clean text from space, formating
df["date_added"] = df["date_added"].str.strip()
df["country"] = df["country"].str.strip()
df["rating"] = df["rating"].str.strip()
```

```python
# fill null values with unknown

df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Unknown")
df.head()
```
